```python
attack_types = {
    'SQL Injection': 'sql injection',
    'Path Traversal': 'path traversal',
    'RCE': 'remote code execution',
    'XSS': 'cross-site scripting',
    'Buffer Overflow': 'buffer overflow',
    'Use-After-Free': 'use-after-free',
    'Authentication Bypass': 'authentication',
    'Privilege Escalation': 'privilege escal',
}
```

```python
vurn['description'].str.lower().str.contains('injection')
```

```python
attack_counts = {'attack_type': [], 'count': []}

for name, keyword in attack_types.items():
    count = vurn['description'].str.lower().str.contains(keyword).sum()
    attack_counts['attack_type'].append(name)
    attack_counts['count'].append(count)

df_attacks = pd.DataFrame(attack_counts).sort_values('count', ascending=False)
```

```python
df_attacks = df_attacks.head(5)
```

```python
sns.barplot(x=df_attacks['attack_type'], y=df_attacks['count'])
plt.title('Top 5 injections by type')
plt.xlabel('Attack Type')
plt.xticks(rotation=45)
```
